## 5. 도미와 빙어 데이터 합치기
## 5. 合并鲷鱼和胡瓜鱼数据

```python
# 리스트끼리 + 하면 이어 붙여진다 : 도미 길이 35개 뒤에 빙어 길이 14개
# 列表相加 + 就是连接：35个鲷鱼长度后面接上14个胡瓜鱼长度
length=bream_length+smelt_length
```

```python
# 도미 무게 35개 뒤에 빙어 무게 14개를 이어 붙인다
# 35个鲷鱼重量后面接上14个胡瓜鱼重量
weight=bream_weight+smelt_weight
```

```python
# 합친 길이 데이터 개수 확인 (35+14=49개)
# 确认合并后的长度数据个数（35+14=49个）
len(length)
```

```python
# 합친 무게 데이터 개수 확인 (49개)
# 确认合并后的重量数据个数（49个）
len(weight)
```

```python
# zip()으로 길이와 무게를 하나씩 짝지어 [길이, 무게] 모양의 2차원 리스트를 만든다
# 用 zip() 把长度和重量一一配对，做成 [长度, 重量] 形式的二维列表
fish_data=[[l,w] for l,w in zip(length, weight)]
```

```python
# 만든 데이터를 확인한다 : 생선 1마리 = [길이, 무게] 1쌍
# 查看生成的数据：1条鱼 = 1对 [长度, 重量]
fish_data
```

```python
# 정답(타깃) 만들기 : 도미는 1로 35개, 빙어는 0으로 14개
# 制作正确答案(目标)：鲷鱼用1表示35个，胡瓜鱼用0表示14个
fish_target=[1]*35 + [0]*14
```

```python
# 정답 리스트를 확인한다
# 查看正确答案列表
fish_target
```

## [추가] KNN은 어떻게 판단할까? (확률과 다수결)
## [补充] KNN 是如何判断的？（概率与多数表决）

```python
# 숫자 계산 도구 numpy를 np라는 이름으로 불러온다
# 以 np 为名导入数值计算工具 numpy
import numpy as np
# 새 생선 데이터 [길이 15, 무게 300]을 만든다
# 创建新鱼数据 [长度15, 重量300]
new_fish = [[15, 300]]
# kneighbors() : 가장 가까운 이웃 5개의 거리(distances)와 위치 번호(indices)를 찾는다
# kneighbors()：找出最近5个邻居的距离(distances)和位置编号(indices)
distances, indices = kn.kneighbors(new_fish, n_neighbors=5)
# indices의 모양을 확인한다 : (1, 5) = 생선 1마리, 이웃 5개
# 查看 indices 的形状：(1, 5) = 1条鱼，5个邻居
print("indices 형태 / 形状:", indices.shape)
# flatten() 으로 1줄로 펴고 tolist() 로 일반 리스트로 바꾼다
# 用 flatten() 展平成一行，再用 tolist() 转成普通列表
indices = indices.flatten().tolist()
# 이웃 5마리의 정답(1=도미, 0=빙어)을 꺼낸다
# 取出5个邻居的答案（1=鲷鱼，0=胡瓜鱼）
neighbor_classes = [fish_target[i] for i in indices]
# 이웃 5마리의 클래스를 출력한다
# 输出5个邻居的类别
print("이웃 5개의 클래스 / 5个邻居的类别:", neighbor_classes)
# 모델이 예측한 클래스를 출력한다 (모델 자체는 이웃 3개 기준)
# 输出模型预测的类别（模型本身以3个邻居为准）
print("예측 클래스 / 预测类别:", kn.predict(new_fish)[0])
# predict_proba() : 각 클래스일 확률을 계산한다
# predict_proba()：计算属于各类别的概率
probabilities = kn.predict_proba(new_fish)[0]
# 클래스 0(빙어)일 확률을 소수점 2자리로 출력한다
# 以两位小数输出属于类别0(胡瓜鱼)的概率
print(f"클래스 0 확률 / 类别0概率: {probabilities[0]:.2f}")
# 클래스 1(도미)일 확률을 출력한다
# 输出属于类别1(鲷鱼)的概率
print(f"클래스 1 확률 / 类别1概率: {probabilities[1]:.2f}")
# count() 로 이웃 중 0과 1이 각각 몇 개인지 센다
# 用 count() 数出邻居中0和1各有几个
class_counts = neighbor_classes.count(0), neighbor_classes.count(1)
# 개수가 더 많은 쪽을 고른다 (다수결)
# 选出个数较多的一方（多数表决）
majority_class = 0 if class_counts[0] > class_counts[1] else 1
# 다수결 결과를 출력한다
# 输出多数表决的结果
print(f"다수결 예측 / 多数表决预测: {majority_class}")
# 클래스 0인 이웃 수
# 类别0的邻居数
print(f"클래스 0 이웃 수 / 类别0邻居数: {class_counts[0]}")
# 클래스 1인 이웃 수
# 类别1的邻居数
print(f"클래스 1 이웃 수 / 类别1邻居数: {class_counts[1]}")
```

## [추가] KNN이 이웃을 빨리 찾는 방법 : KD-트리
## [补充] KNN 快速寻找邻居的方法：KD-树

- `fit()` 을 실행하면 사이킷런은 데이터를 저장하면서, 이웃을 빨리 찾기 위한 **트리 구조**를 만든다.
- 执行 `fit()` 时，sklearn 会保存数据，同时建立用于快速寻找邻居的**树结构**。
- KD-트리는 데이터를 **x축 기준 → y축 기준 → x축 기준 …** 으로 번갈아 반씩 나눈다.
- KD-树按照 **x轴 → y轴 → x轴 …** 的顺序交替把数据对半分割。
- 아래 그림은 처음 2단계(세로선 1개 → 가로선 2개)만 그린 것이다.
- 下图只画出了前2步（1条竖线 → 2条横线）。

```python
# numpy와 matplotlib를 불러온다
# 导入 numpy 和 matplotlib
import numpy as np
import matplotlib.pyplot as plt
# 생선 데이터(X)와 정답(y)을 배열로 준비한다
# 把鱼的数据(X)和答案(y)准备成数组
X = np.array(fish_data)
y = np.array(fish_target)
# [1단계] 길이(x축)의 중앙값으로 데이터를 왼쪽/오른쪽 반으로 나눈다
# [第1步] 以长度(x轴)的中位数把数据分成左右两半
split_x = np.median(X[:, 0])
# 왼쪽 그룹(길이가 중앙값보다 작은 생선)
# 左侧组（长度小于中位数的鱼）
left = X[X[:, 0] < split_x]
# 오른쪽 그룹(길이가 중앙값 이상인 생선)
# 右侧组（长度大于等于中位数的鱼）
right = X[X[:, 0] >= split_x]
# [2단계] 각 그룹을 무게(y축)의 중앙값으로 다시 위/아래로 나눈다
# [第2步] 再把每组以重量(y轴)的中位数分成上下两半
split_left_y = np.median(left[:, 1])
split_right_y = np.median(right[:, 1])
# 그래프 크기를 정한다
# 设置图表大小
plt.figure(figsize=(12, 10))
# 빙어(0)는 빨간 점, 도미(1)는 파란 점으로 그린다
# 胡瓜鱼(0)画红点，鲷鱼(1)画蓝点
plt.scatter(X[y==0, 0], X[y==0, 1], c='red', label='Class 0 (Smelt)')
plt.scatter(X[y==1, 0], X[y==1, 1], c='blue', label='Class 1 (Bream)')
# 1단계 나눔선 : 빨간 세로선
# 第1步分割线：红色竖线
plt.axvline(split_x, color='r', label='Split 1 (Length)')
# 2단계 나눔선 : 왼쪽/오른쪽 영역에 파란 가로선
# 第2步分割线：在左/右区域画蓝色横线
plt.plot([X[:, 0].min(), split_x], [split_left_y, split_left_y], 'b-', label='Split 2 (Weight)')
plt.plot([split_x, X[:, 0].max()], [split_right_y, split_right_y], 'b-')
# 제목과 축 이름
# 标题和坐标轴名称
plt.title('KD-Tree Visualization for Fish Data (first 2 splits)')
plt.xlabel('Length (cm)')
plt.ylabel('Weight (g)')
# 범례 표시
# 显示图例
plt.legend()
# 그래프 보여주기
# 显示图表
plt.show()
```

```python
# numpy와 matplotlib를 불러온다
# 导入 numpy 和 matplotlib
import numpy as np
import matplotlib.pyplot as plt
# 매번 같은 난수가 나오도록 seed를 0으로 고정한다
# 把随机种子固定为0，使每次生成相同的随机数
np.random.seed(0)
# 0~1 사이의 임의의 점 20개(x, y)를 만든다
# 生成20个0~1之间的随机点(x, y)
X = np.random.rand(20, 2)
# [1단계] x값의 중앙값으로 세로로 나눈다
# [第1步] 以x值的中位数纵向分割
split_x = np.median(X[:, 0])
left = X[X[:, 0] < split_x]
right = X[X[:, 0] >= split_x]
# [2단계] 왼쪽/오른쪽 그룹을 각각 y값의 중앙값으로 가로로 나눈다
# [第2步] 左/右两组分别以y值的中位数横向分割
split_left_y = np.median(left[:, 1])
split_right_y = np.median(right[:, 1])
# 그래프 크기를 정하고 점을 초록색으로 그린다
# 设置图表大小，并把点画成绿色
plt.figure(figsize=(10, 10))
plt.scatter(X[:, 0], X[:, 1], c='g')
# 빨간 세로선(1단계), 파란 가로선(2단계)을 그린다
# 画红色竖线(第1步)和蓝色横线(第2步)
plt.plot([split_x, split_x], [0, 1], 'r-')
plt.plot([0, split_x], [split_left_y, split_left_y], 'b-')
plt.plot([split_x, 1], [split_right_y, split_right_y], 'b-')
# 제목과 축 이름
# 标题和坐标轴名称
plt.title('KD-Tree Visualization (first 2 splits)')
plt.xlabel('X')
plt.ylabel('Y')
# 그래프 보여주기
# 显示图表
plt.show()
```
